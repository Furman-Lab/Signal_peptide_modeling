```python
# function to parse the table outputs of SignalP6
def read_and_parse_signalp_tabs(tab):
    signalp_tab = pd.read_table(tab, skiprows=1)
    signalp_tab.columns = signalp_tab.columns.str.replace('#','').str.replace(' ','_')
    return signalp_tab
```
